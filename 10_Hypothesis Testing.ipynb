### What is Hypothesis Testing?
Hypothesis Testing is a statistical method used to determine whether there is enough evidence to accept or reject a claim about a population.

Simple Example

Suppose a company claims:

"The average delivery time is 30 minutes."

You collect data from 100 deliveries and want to check whether this claim is true.

```python
# sample dataset

sample = [9.5, 10.2, 9.8, 10.1, 9.7, 10.3, 9.9, 10.0, 9.6, 10.1]
```

### Null Hypothesis (H₀)

The Null Hypothesis is the default assumption or the statement that there is no significant change, difference, or effect.

Symbol: H₀

Example

Company claims average delivery time is 30 minutes:

$$ H_0: \mu = 30 $$

Meaning:

"The population average delivery time is 30 minutes."

Easy way to remember

H₀ = Nothing changed / No effect / No difference

### Alternative Hypothesis (H₁ or Hₐ)

The Alternative Hypothesis is what we are trying to find evidence for.

Symbol: H₁ or Hₐ

Example

If you believe the average delivery time is different from 30 minutes:

$$ H_a: \mu \neq 30 $$

Meaning:

"The average delivery time is not 30 minutes."

Easy way to remember

Hₐ = Something changed / There is an effect / There is a difference

| Hypothesis | Meaning           |
| ---------- | ----------------- |
| H₀: μ = 30 | Average is 30     |
| Hₐ: μ ≠ 30 | Average is not 30 |


### p-value

The p-value tells us how much evidence we have against the Null Hypothesis.

More technically:

The p-value is the probability of observing a result at least as extreme as the sample result, assuming H₀ is true.

Common rule

Usually:

α=0.05

| p-value  | Decision          |
| -------- | ----------------- |
| p ≤ 0.05 | Reject H₀         |
| p > 0.05 | Fail to reject H₀ |


```python
p_value = 0.03

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

### Example
Suppose:

p-value = 0.02
α = 0.05

Since:

$$ 0.02 < 0.05 $$

We reject H₀.

There is statistically significant evidence against the null hypothesis.

Important

Don't say:

"p-value = probability that H₀ is true."

That's incorrect.

### Confidence Interval

A Confidence Interval (CI) gives a range of values within which the population parameter is estimated to lie.

For example:

Average salary = ₹50,000
95% CI = ₹48,000 – ₹52,000

We can say:

The estimated population average is between ₹48,000 and ₹52,000 at the 95% confidence level.

Formula

For a mean, a common form is:

$$ CI = \bar{x} \pm Critical\ Value \times SE $$

where:

* \(\bar{x}\) = sample mean
* SE = standard error
Example

Suppose:

Sample mean = 100
95% CI = [96, 104]

The interval tells us the plausible range for the population mean.

Relationship with hypothesis testing

For a two-tailed test at α = 0.05, a 95% confidence interval corresponds to the same significance threshold.

### Significance Level (α)

The Significance Level is the threshold we choose for deciding whether a result is statistically significant.

Symbol:

$$ \alpha $$

Common values:

α = 0.05  →  5%
α = 0.01  →  1%
α = 0.10  → 10%
Example

If:

α = 0.05
p-value = 0.03

Then:

$$ 0.03 < 0.05 $$

So we reject H₀.

Simple meaning

α = maximum probability of Type I Error that we are willing to accept.

### Type I Error

A Type I Error happens when we reject H₀ even though H₀ is actually true.

In simple words:

False Positive

Example

Suppose:

H₀: Person does NOT have a disease

Test says:

Reject H₀ → Person has disease

But actually the person is healthy.

That's a Type I Error.

Memory trick

Type I = False Positive

### Type II Error

A Type II Error happens when we fail to reject H₀ even though H₀ is actually false.

In simple words:

False Negative

Example
H₀: Person does NOT have a disease

Test says:

Fail to reject H₀ → Person does not have disease

But actually the person has the disease.

That's a Type II Error.

Memory trick

Type II = False Negative

Easy memory

Type I → False Positive;
Type II → False Negative

| Reality  | Our Decision      | Result            |
| -------- | ----------------- | ----------------- |
| H₀ true  | Reject H₀         | **Type I Error**  |
| H₀ true  | Fail to reject H₀ | Correct           |
| H₀ false | Reject H₀         | Correct           |
| H₀ false | Fail to reject H₀ | **Type II Error** |


### One-Tailed Test

A one-tailed test checks for an effect in one specific direction.

There are two types:

Right-tailed test

Testing whether something is greater than a value.

Example:

$$ H_0: \mu = 100 $$ $$ H_a: \mu > 100 $$

Question:

Is the average greater than 100?

Left-tailed test

Testing whether something is less than a value.

$$ H_0: \mu = 100 $$ $$ H_a: \mu < 100 $$

Question:

Is the average less than 100?

Example

A machine is supposed to produce 100 units/hour.

You want to check whether it produces more than 100:

$$ H_0: \mu = 100 $$ $$ H_a: \mu > 100 $$

This is a right-tailed test.

### Two-Tailed Test

A two-tailed test checks whether the value is different in either direction.

Example:

$$ H_0: \mu = 100 $$ $$ H_a: \mu \neq 100 $$

We are asking:

Is the average different from 100?

It could be:

90
95
105
110

Both directions matter.

Example

A company says:

"Average customer waiting time is 10 minutes."

You want to know whether it is different from 10 minutes.

Therefore:

$$ H_0: \mu = 10 $$ $$ H_a: \mu \neq 10 $$

This is a two-tailed test.

| Feature          | One-Tailed           | Two-Tailed      |
| ---------------- | -------------------- | --------------- |
| Direction        | One direction        | Both directions |
| Alternative      | `>` or `<`           | `≠`             |
| Example          | μ > 100              | μ ≠ 100         |
| Checks           | Increase OR decrease | Any difference  |
| Critical regions | One tail             | Two tails       |


| Concept                         | Simple Meaning                             |
| ------------------------------- | ------------------------------------------ |
| **Null Hypothesis (H₀)**        | No effect / no difference                  |
| **Alternative Hypothesis (Hₐ)** | Effect / difference exists                 |
| **p-value**                     | Evidence against H₀                        |
| **Confidence Interval**         | Plausible range for a population parameter |
| **Significance Level (α)**      | Threshold for statistical significance     |
| **Type I Error**                | False Positive                             |
| **Type II Error**               | False Negative                             |
| **One-Tailed Test**             | Tests one direction                        |
| **Two-Tailed Test**             | Tests both directions                      |

* H₀       → No change
* Hₐ       → Change exists

* p ≤ α    → Reject H₀
* p > α    → Fail to reject H₀

* Type I   → False Positive
* Type II  → False Negative

* >    → Right-tailed
* <        → Left-tailed
* ≠        → Two-tailed
