### Probability
Probability is the measure of how likely an event is to happen.

The value of probability always lies between 0 and 1.

0 → Impossible event
1 → Certain event
Formula:

P(E)= Number of Favorable Outcomes / Total Number of Outcomes

Example:

Consider rolling a fair six-sided dice.

Possible outcomes: {1, 2, 3, 4, 5, 6}

Probability of getting 4 : P(4)=1/6

```python
favorable = 3
total = 6

probability = favorable / total

print(probability)
```

### Independent Events

Two events are independent when the occurrence of one event does not affect the other.
For independent events:
P(A∩B)=P(A)P(B)

Example

Toss a coin twice.

Event A = first toss is Head.

Event B = second toss is Head.

The first toss doesn't affect the second toss.

Therefore they are independent.

$$ P(A)=\frac12 $$ $$ P(B)=\frac12 $$ $$ P(A\cap B)=\frac12\times\frac12=\frac14 $$

So the probability of getting Head twice = 25%.

```python
import random

dice = random.randint(1,6)
coin = random.choice(["Head","Tail"])

print(dice)
print(coin)
```

### Dependent Events
Two events are Dependent if the occurrence of one event affects the probability of the other.

Example:

A bag contain,

3 Red Balls
2 Blue Balls
If one red ball is removed without replacement, the probability of drawing another red ball changes.

| Independent                      | Dependent                         |     |
| -------------------------------- | --------------------------------- | --- |
| One event doesn't affect another | One event affects another         |     |
| Coin tosses                      | Drawing cards without replacement |     |
| `P(A∩B)=P(A)P(B)`                | `P(A∩B)=P(A)P(B                   | A)` |


```python
balls = ["R","R","R","B","B"]

first = balls.pop(0)

print("Remaining:", balls)
```

### Conditional Probability

Conditional probability tells us the probability of an event given that another event has already happened.

Notation:

$$ P(A|B) $$

Read as:

Probability of A given B.

Formula:

P(A∣B) = P(A∩B) / P(B)

Example:

A dice is rolled.

Given that the number is even, what is the probability that it is greater than 4?

Even numbers : {2,4,6}

Greater than 4 : {6}

Probability : 1/3

```python
boys = 60
boys_who_know_python = 30

probability = boys_who_know_python / boys

print(probability)
```

### Bayes' Theorem

Bayes' theorem is one of the most important probability concepts in AI, ML, statistics, and data science.
Formula: P(A∣B) = (P(B∣A)×P(A)) / P(B)

Example:

Suppose

P(Disease)=0.01

P(Positive|Disease)=0.99

P(Positive)=0.05

Then

P(Disease|Positive)

=(0.99×0.01)/0.05

=0.198

### Expected Value
Expected value is the long-run average outcome of a random variable.

It can be thought of as a probability-weighted average.
| Outcome | Probability |
| ------: | ----------: |
|    ₹100 |         0.5 |
|    ₹200 |         0.3 |
|    ₹500 |         0.2 |

E(X)=100(0.5)+200(0.3)+500(0.2)
$$ =50+60+100 $$
=₹210

```python
values = [100, 200, 500]
probabilities = [0.5, 0.3, 0.2]

expected_value = sum(
    value * probability
    for value, probability in zip(values, probabilities)
)

print(expected_value)
```
