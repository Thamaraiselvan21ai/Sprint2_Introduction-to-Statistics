### Sampling
Sampling is the process of selecting a smaller group of people, objects, or data points from a larger population to study and make conclusions about the entire population.
| Term           | Meaning                                |
| -------------- | -------------------------------------- |
| **Population** | Entire group we are interested in      |
| **Sample**     | Smaller group selected from population |
| **Sampling**   | Process of selecting the sample        |
| **Parameter**  | Numerical value describing population  |
| **Statistic**  | Numerical value calculated from sample |


```python
import random

students = list(range(1, 1001))

sample = random.sample(students, 100)
print(sample[:10])   # Display first 10 selected students
```

### Random Sampling
Random sampling means selecting samples randomly so that every member of the population has an equal chance of being selected.

Example

Population:

Students = [A, B, C, D, E, F, G, H, I, J]

Randomly select 3:
Sample = [B, G, I]

```python
import random

students = ["A", "B", "C", "D", "E",
            "F", "G", "H", "I", "J"]

sample = random.sample(students, 3)

print(sample)
```

### Advantages
* Simple
* Reduces selection bias
* Each member has a chance of selection
#### Disadvantage

It can be difficult when the population is very large or geographically distributed.

### Stratified Sampling

Stratified sampling divides the population into different groups called strata, and then randomly selects samples from each group.

The groups are created based on an important characteristic.
Example:

The college has students from:

First Year
Second Year
Third Year
Fourth Year

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["A", "B", "C", "D", "E", "F"],
    "department": ["IT", "IT", "HR", "HR", "Finance", "Finance"]
})

sample = df.groupby("department", group_keys=False).apply(
    lambda x: x.sample(1)
)

print(sample)
```

```python

first_year = list(range(1, 251))
second_year = list(range(251, 501))
third_year = list(range(501, 751))
fourth_year = list(range(751, 1001))

import random

sample = (
    random.sample(first_year, 25) +
    random.sample(second_year, 25) +
    random.sample(third_year, 25) +
    random.sample(fourth_year, 25)
)

print(sample[:10])
```

### Cluster Sampling
Cluster Sampling divides the population into clusters, then randomly selects one or more entire clusters for the study.

Example:

The college has 20 classrooms.

The principal randomly selects 2 classrooms and surveys every student in those classrooms.

#### Advantage
* Cheaper
* Easier for geographically distributed populations
### Disadvantage

Clusters may not perfectly represent the entire population.

```python
import random

clusters = ["Class A", "Class B", "Class C", "Class D", "Class E"]

selected_cluster = random.choice(clusters)
print(selected_cluster)
```

### Systematic Sampling

Systematic sampling selects every k-th member from an ordered population after choosing a random starting point.

Formula

If:

Population size = N
Sample size = n

Then:

$$ k = \frac{N}{n} $$
Example

Population:

10,000 customers

Required sample:

1,000 customers

Therefore:

k=
10000/
1000
	​

=10

```python
students = list(range(1, 1001))

systematic_sample = students[::10]

print(systematic_sample[:10])
#Students are selected at regular intervals.
```

### Sampling Bias
Sampling Bias occurs when some members of the population have a higher chance of being selected than others, making the sample unrepresentative.

Example:

The principal surveys only final-year students about the canteen.

The opinions of first-, second-, and third-year students are ignored.

```python
final_year_students = list(range(751, 1001))
print(final_year_students[:10])
#The sample does not represent the entire student population.
```

### Sampling Error
Sampling Error is the difference between the results obtained from a sample and the actual population value.

Common Example

Suppose:

Population average satisfaction = 8.5
Sample average satisfaction = 8.1
The difference (0.4) is the sampling error.


```python
population_mean = 8.5
sample_mean = 8.1

sampling_error = population_mean - sample_mean

print(sampling_error)
```

### Sample Size

Sample size is the number of observations or individuals included in a sample.

Example:

Population = 100,000 customers
Sample = 2,000 customers

Sample size = 2,000

### Factors Affecting Sample Size
#### 1. Population Size

Larger populations may require larger samples, although sample size does not need to grow proportionally.

#### 2. Confidence Level

Common confidence levels:

90%
95%
99%

Higher confidence generally requires a larger sample.

#### 3. Margin of Error

Smaller margin of error requires a larger sample.

For example:

±5%  → smaller sample
±2%  → larger sample
#### 4. Population Variability

More variation in the population generally requires a larger sample.

Basic Sample Size Formula

For estimating a population proportion:

$$ n = \frac{Z^2p(1-p)}{E^2} $$

Where:

Symbol	Meaning
n	Required sample size
Z	Z-score for confidence level
p	Estimated population proportion
E	Margin of error

For 95% confidence:

$$ Z = 1.96 $$

If we don't know p, a common conservative choice is:

$$ p = 0.5 $$
Example

Suppose:

Confidence = 95%
Margin of error = 5%
p = 0.5

Then:

$$ n = \frac{1.96^2(0.5)(0.5)}{0.05^2} $$

Approximately:

$$ n \approx 385 $$

So around 385 observations are needed under these assumptions.
