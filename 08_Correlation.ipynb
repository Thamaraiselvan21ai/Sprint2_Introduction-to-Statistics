### Correlation

Correlation measures the strength and direction of the relationship between two variables.

For example:

Study hours ↑ → Exam marks ↑ → Positive correlation
Price ↑ → Demand ↓ → Negative correlation
Shoe size and exam marks → usually No correlation

The correlation coefficient normally ranges from -1 to +1.

| Correlation value | Meaning                      |
| ----------------: | ---------------------------- |
|                +1 | Perfect positive correlation |
|     +0.7 to +0.99 | Strong positive              |
|     +0.3 to +0.69 | Moderate positive            |
|                 0 | No linear correlation        |
|     -0.3 to -0.69 | Moderate negative            |
|     -0.7 to -0.99 | Strong negative              |
|                -1 | Perfect negative             |


```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "Marks": [45, 50, 55, 65, 70, 75, 85, 90]
})

print(data)
```

### Covariance

Covariance tells us whether two variables tend to increase or decrease together.

Formula:

$$ Cov(X,Y)=\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{n-1} $$

For a sample:
| Covariance | Meaning                                         |
| ---------: | ----------------------------------------------- |
|   Positive | X and Y tend to increase together               |
|   Negative | One tends to increase while the other decreases |
|     Near 0 | Little/no linear relationship                   |


| Covariance               | Correlation                |
| ------------------------ | -------------------------- |
| Shows direction          | Shows direction + strength |
| Value has no fixed range | Always between -1 and +1   |
| Depends on units         | Unitless                   |
| Harder to interpret      | Easier to interpret        |


### Positive Correlation
A Positive Correlation means that when one variable increases, the other variable also increases.

Example:

Students who study more generally score higher marks.
X ->Increases Y -> Increases

```python
print(data["Study_Hours"].corr(data["Marks"]))
```

### Negative Correlation

When X increases, Y tends to decrease.

Example:

Product price ↑ → Demand ↓


```python
import pandas as pd

data2 = pd.DataFrame({
    "TV_Hours":[1,2,3,4,5,6,7,8],
    "Marks":[90,85,80,75,70,65,60,55]
})

print(data2.corr()) #More TV hours → Lower marks.
```

### No Correlation
No Correlation means there is no relationship between two variables.

Example:

A student's roll number has no relationship with exam marks.

```python
import pandas as pd

data3 = pd.DataFrame({
    "Roll_No":[1,2,3,4,5,6,7,8],
    "Marks":[75,60,85,55,90,70,65,80]
})

print(data3.corr()) #Changing the roll number does not affect marks.
```

### Pearson Correlation
Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.

It is represented by:

$$ r $$

Formula:

$$ r = \frac{Cov(X,Y)} {\sigma_X\sigma_Y} $$

Where:

* Cov(X,Y) = covariance
* σX = standard deviation of X
* σY = standard deviation of Y
The value ranges from -1 to +1.

Example:

The school measures the relationship between study hours and marks.

```python
import pandas as pd

data = {
    "Hours": [2, 4, 6, 8, 10],
    "Marks": [40, 50, 65, 80, 90]
}

df = pd.DataFrame(data)

correlation = df["Hours"].corr(df["Marks"])

print(correlation) #A value close to +1 indicates a strong positive linear relationship.
```

### Spearman Correlation
Spearman Correlation measures the relationship between the rank (order) of two variables.

It is useful when the relationship is not perfectly linear.

Example:

The school ranks students based on study hours and exam marks.

```python
spearman = df["Hours"].corr(df["Marks"], method="spearman")

print(spearman)
# A value close to +1 means that students with higher study-hour ranks also have higher mark ranks.
```

| Feature                           | Pearson             | Spearman                 |
| --------------------------------- | ------------------- | ------------------------ |
| Measures                          | Linear relationship | Monotonic relationship   |
| Uses                              | Actual values       | Ranks                    |
| Outlier sensitivity               | More sensitive      | Generally less sensitive |
| Suitable for ordinal data         | No                  | Yes                      |
| Non-linear monotonic relationship | Not ideal           | Good                     |
