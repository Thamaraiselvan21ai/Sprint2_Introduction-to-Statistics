### Data Distribution
A data distribution describes how data values are spread or arranged. It shows whether the data is concentrated around a central value, spread evenly, or skewed to one side.

Understanding data distribution helps us choose the right statistical methods and machine learning algorithms.

### Normal Distribution
Definition

A Normal Distribution is a continuous probability distribution where most values are concentrated around the mean, and fewer values occur as we move away from the mean.

It looks like a bell-shaped curve.

Mean = Median = Mode

Most values lie near the center.

Fewer values occur as we move away from the mean.

#### Important properties
* Bell-shaped
* Symmetric
* Mean = Median = Mode
* Most observations are near the mean
* Tails extend in both directions
* Total area under the curve = 1

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample marks of 100 students
marks = np.random.normal(loc=70, scale=10, size=100)

plt.hist(marks, bins=10, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

Most students scored around the average, and fewer students scored very low or very high marks.

### Uniform Distribution
A Uniform Distribution occurs when every value within a specified range has an equal probability of occurring.
For example, a fair dice has equal probability for:

1, 2, 3, 4, 5, 6

Each number has probability:

$$ P(X=x)=\frac{1}{6} $$
Shape

Unlike normal distribution, uniform distribution is approximately flat.

```python
uniform_marks = np.random.uniform(0, 100, 100)

plt.hist(uniform_marks, bins=10, edgecolor="black")
plt.title("Uniform Distribution")
plt.show()

#Students are evenly distributed across all mark ranges.
```

### Gaussian Distribution
A Gaussian Distribution is another name for the Normal Distribution. The terms are often used interchangeably because the distribution was described by the mathematician Carl Friedrich Gauss.
Formula

The probability density function is:

$$ f(x)= \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$

Where:

μ = mean
σ = standard deviation
x = observation

```python
import numpy as np

data = np.random.normal(0, 1, 1000)
print(data[:10])
```

### AI/ML Example

Gaussian assumptions appear in:

* Gaussian Naive Bayes
* Gaussian Mixture Models
* Kalman filters
* Statistical modeling
* Noise modeling

### Left Skew (Negative Skew)
A Left-Skewed Distribution has a long tail on the left side.

Most observations are concentrated on the right side.
Relationship

Generally:

Mean<Median<Mode

```python
left_skew = np.random.beta(5, 2, 100) * 100

plt.hist(left_skew, bins=10, edgecolor="black")
plt.title("Left Skewed Distribution")
plt.show()
```

### Right Skew (Positive Skew)
A Right-Skewed Distribution has a long tail on the right side.

Most observations are concentrated on the left side.
Relationship

Generally:

Mode<Median<Mean

```python
right_skew = np.random.beta(2, 5, 100) * 100

plt.hist(right_skew, bins=10, edgecolor="black")
plt.title("Right Skewed Distribution")
plt.show()
```

### Symmetric Distribution
A Symmetric Distribution has identical left and right halves when divided through the center.
Equal spread on both sides.
Important point
For a perfectly symmetric unimodal distribution:
Mean=Median=Mode

```python
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
#The left and right sides of the distribution are nearly equal.
```

### Kurtosis
Kurtosis measures how heavy or light the tails of a distribution are compared to a normal distribution. It indicates the presence of extreme values (outliers).

Types:

* Mesokurtic – Similar to a normal distribution.
* Leptokurtic – Sharper peak with heavier tails (more outliers).
* Platykurtic – Flatter peak with lighter tails (fewer outliers).

```python
import pandas as pd

data = pd.Series([10, 12, 13, 14, 15, 16, 17, 18, 20])

print(data.kurt())
```

| Type        | Tails       | Excess Kurtosis |
| ----------- | ----------- | --------------: |
| Mesokurtic  | Normal-like |               0 |
| Leptokurtic | Heavy       |             > 0 |
| Platykurtic | Light       |             < 0 |


### Empirical Rule — 68–95–99.7 Rule

The Empirical Rule applies to approximately normal distributions.

It tells us how much data falls within a certain number of standard deviations from the mean.
* About 68% of observations lie within 1 standard deviation of the mean.
* About 95% lie within 2 standard deviations.
* About 99.7% lie within 3 standard deviations.

Within 1 standard deviation

Approximately:

$$ 68\% $$

of data lies between:

$$ \mu-\sigma $$

and

$$ \mu+\sigma $$
Within 2 standard deviations

Approximately:

$$ 95\% $$

of data lies between:

$$ \mu-2\sigma $$

and

$$ \mu+2\sigma $$
Within 3 standard deviations

Approximately:

$$ 99.7\% $$

of data lies between:

$$ \mu-3\sigma $$

and

μ+3σ

```python

import numpy as np

data = np.random.normal(70, 10, 1000)

mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("1 SD:", mean - std, "to", mean + std)
print("2 SD:", mean - 2*std, "to", mean + 2*std)
print("3 SD:", mean - 3*std, "to", mean + 3*std)
```
