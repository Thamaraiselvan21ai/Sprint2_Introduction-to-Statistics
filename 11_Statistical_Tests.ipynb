### Statistical Tests
A company wants to analyze employee salaries and determine whether there are significant differences between departments or groups.
| Test                | Main purpose                                                           | Typical data           |
| ------------------- | ---------------------------------------------------------------------- | ---------------------- |
| **T-Test**          | Compare means                                                          | Numerical              |
| **Z-Test**          | Compare mean/proportion when population SD is known or sample is large | Numerical / proportion |
| **Chi-Square Test** | Test relationships between categorical variables                       | Categorical            |
| **ANOVA**           | Compare means of 3+ groups                                             | Numerical              |
| **Mann-Whitney U**  | Compare 2 independent groups without normality assumption              | Ordinal/numerical      |
| **Shapiro-Wilk**    | Test whether data is normally distributed                              | Numerical              |


```python
# sample dataset

import pandas as pd

data = {
    "Salary": [45000, 47000, 46000, 48000, 50000,
               52000, 53000, 51000, 54000, 55000],
    "Department": ["HR","HR","HR","HR","HR",
                   "IT","IT","IT","IT","IT"],
    "Gender": ["Male","Female","Male","Female","Male",
               "Female","Male","Female","Male","Female"]
}

df = pd.DataFrame(data)

print(df)
```

### T-Test

A T-Test is used to determine whether the means of two groups are significantly different.

Example

Suppose:

Average sales before advertising = ₹50,000
Average sales after advertising = ₹55,000

A t-test helps answer:

"Is this ₹5,000 difference statistically significant, or could it have happened by chance?"

### Types
One-sample t-test → Compare one sample mean with a known value.
*Independent two-sample t-test → Compare two independent groups.
Paired t-test → Compare the same subjects before and after.
### Assumptions
* Data is numerical.
* Observations are independent for an independent t-test.
* Data is approximately normally distributed.
* For the standard Student's independent t-test, the two groups should have similar variances.
* No extreme outliers.

### When to use

### Use a t-test when:

* Comparing means.
* Sample size is relatively small.
* Population standard deviation is unknown.
* Data is approximately normally distributed.

Example

### Compare average salary of:

* Male employees
* Female employees
### When NOT to use

#### Don't use it when:

* Data is strongly non-normal with small samples.
* You have 3 or more independent groups → use ANOVA.
* Data is categorical → consider Chi-Square.
* You want to compare distributions/ranks and normality is inappropriate → consider Mann-Whitney U.

```python
#Compare the average salaries of HR and IT employees
from scipy.stats import ttest_ind

hr = [45000,47000,46000,48000,50000]
it = [52000,53000,51000,54000,55000]

t_stat, p_value = ttest_ind(hr, it)

print("T Statistic:", t_stat)
print("P Value:", p_value)
```

p < 0.05 → Significant difference between HR and IT salaries;  
p ≥ 0.05 → No significant difference.

### Z-Test

A Z-Test is used to determine whether a sample mean or proportion is significantly different from a hypothesized population value.

It is closely related to the t-test.

| T-Test                        | Z-Test                                             |
| ----------------------------- | -------------------------------------------------- |
| Population SD usually unknown | Population SD known, or large-sample approximation |
| Uses t-distribution           | Uses normal/Z distribution                         |
| Common for smaller samples    | Common for large samples                           |

Assumptions
* Data is independent.
* Data is approximately normally distributed for small samples.
* Population standard deviation is known or sample is sufficiently large.
* For a proportion z-test, observations should satisfy appropriate large-sample conditions.

### When to use

Example:

A manufacturer claims:

Average battery life = 10 hours.

You collect a large sample and want to test this claim.

### When NOT to use

Don't use a z-test when:

* Population SD is unknown and sample is small.
* Data strongly violates assumptions.
* You are comparing 3+ means → ANOVA.
* Variables are categorical → Chi-Square.

For many practical small-sample mean problems, t-test is preferred.

### Business Example

A company claims its delivery time is:

30 minutes on average

You collect 1000 delivery records and test whether the actual average is different from 30 minutes.

```python
import numpy as np

sample = np.array([45000,47000,46000,48000,50000])

population_mean = 50000
population_std = 4000

z = (sample.mean() - population_mean) / (population_std / np.sqrt(len(sample)))

print("Z Statistic:", z)
#A large positive or negative Z-score indicates that the sample mean differs considerably from the population mean.
```

### Chi-Square Test

The Chi-Square Test is mainly used with categorical data.

It checks whether observed frequencies differ significantly from expected frequencies or whether two categorical variables are associated.

Two common types
* Chi-Square Goodness-of-Fit
* Chi-Square Test of Independence

#### Assumptions
* Data is categorical/count data.
* Observations are independent.
* Expected frequencies should generally not be too small.
* Each observation belongs to one category.
When to use

Use Chi-Square when investigating:

categorical variable ↔ categorical variable

Example:

* Gender ↔ Product preference
* Education ↔ Loan approval
* Region ↔ Customer churn
  
### When NOT to use

Don't use it when:

* Your main variables are continuous measurements.
* Expected frequencies are very small.
* Observations are not independent.

For very small expected counts, Fisher's exact test may be more appropriate for a 2×2 table.

```python
from scipy.stats import chi2_contingency

data = [
    [60, 40],
    [40, 60]
]

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square:", chi2)
print("P-value:", p_value)
print("Expected:", expected)

#p < 0.05 → Variables are associated.
#p ≥ 0.05 → Variables are independent.
```

### ANOVA (Analysis of Variance)
ANOVA compares the means of three or more groups.

#### Assumptions:

* Normal distribution.
* Equal variances.
* Independent observations.
### When to Use:

* Compare three or more groups.
### When Not to Use:

* Only two groups (use T-Test).
#### Example:

Compare salaries among HR, IT, and Sales departments.



```python
from scipy.stats import f_oneway

hr = [45000,47000,46000]
it = [52000,53000,51000]
sales = [48000,50000,49000]

f_stat, p_value = f_oneway(hr, it, sales)

print("P Value:", p_value)

#p < 0.05 → At least one group mean is significantly different.
#p ≥ 0.05 → No significant difference among the groups.
```

### Mann-Whitney U Test
The Mann-Whitney U Test compares two independent groups when the data is not normally distributed.

### Assumptions:

* Two independent groups.
* Ordinal or continuous data.
* No normality assumption.
### When to Use:

* Non-normal data.
* Small sample size.
### When Not to Use:

* More than two groups.
* Paired observations.
Example:

Compare HR and IT salaries when salary data is not normally distributed.

```python
from scipy.stats import mannwhitneyu

hr = [45000,47000,46000,48000,50000]
it = [52000,53000,51000,54000,55000]

u_stat, p_value = mannwhitneyu(hr, it)

print("P Value:", p_value)
```

#p < 0.05 → Groups differ significantly.

#p ≥ 0.05 → No significant difference.

### Shapiro-Wilk Test
The Shapiro-Wilk Test checks whether data follows a normal distribution.

### Assumptions:

* Continuous data.
* Independent observations.
### When to Use:

Before choosing between parametric and non-parametric tests.
### When Not to Use:

Very large datasets (other normality tests or visual methods may be preferred).
Example:

Check whether employee salary data is normally distributed before performing a T-Test.

```python
from scipy.stats import shapiro

salary = [45000,47000,46000,48000,50000,
          52000,53000,51000,54000,55000]

stat, p_value = shapiro(salary)

print("P Value:", p_value)
```

### p > 0.05 → Data is normally distributed.

### p ≤ 0.05 → Data is not normally distributed.
