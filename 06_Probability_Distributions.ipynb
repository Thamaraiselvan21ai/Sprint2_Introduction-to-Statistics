### Probability Distributions

Probability distributions describe how likely different outcomes are for a random variable.

A simple way to remember:

* Bernoulli → One trial, two outcomes
* Binomial → Multiple Bernoulli trials
* Poisson → Number of events in a fixed interval
* Uniform → Every value has equal probability
* Normal → Bell-shaped distribution around the mean
* Exponential → Waiting time between events

### Bernoulli Distribution

The Bernoulli distribution is used when there is only one trial and the result has exactly two possible outcomes:

Success → 1
Failure → 0
Formula
$$ P(X=x)=p^x(1-p)^{1-x} $$

Where:

\(X\) = outcome
\(p\) = probability of success
\(1-p\) = probability of failure
\(x\) = 0 or 1
Example

Suppose a customer either:

Buys a product → 1
Does not buy → 0

If probability of purchase is 0.3:

$$ P(X=1)=0.3 $$ $$ P(X=0)=0.7 $$
Where is it used?
Coin toss
Customer purchase: Yes/No
Loan default: Yes/No
Disease detection: Positive/Negative
Email: Spam/Not Spam
AI/ML Example

Binary classification often has a Bernoulli-type target:

Customer → Will buy?
             ↓
          Yes / No

For example:

y = [1, 0, 1, 1, 0, 0]

where:

1 = customer buys
0 = customer doesn't buy

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.3

x = [0, 1]
probability = [1-p, p]

plt.bar(x, probability)
plt.xticks([0, 1], ["Failure", "Success"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

### Binomial Distribution
The Binomial distribution tells us the probability of getting a certain number of successes in a fixed number of independent trials.

It is basically:

Bernoulli repeated multiple times.

Formula
$$ P(X=k)=\binom{n}{k}p^k(1-p)^{n-k} $$

Where:

\(n\) = number of trials
\(k\) = number of successes
\(p\) = probability of success
\(1-p\) = probability of failure

#### Where is it used?
* Number of successful sales
* Number of defective products
* Number of customers responding to an advertisement
* Number of correct answers
* A/B testing
* AI/ML Example

Suppose an ML model makes predictions on 100 independent samples.

If probability of correct prediction is 0.9:

n = 100
p = 0.90

Binomial distribution can model the number of correct predictions.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.3

x = np.arange(0, n + 1)
probability = binom.pmf(x, n, p)

plt.bar(x, probability)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

### Poisson Distribution

The Poisson distribution is used to calculate the probability of a certain number of events occurring within a fixed interval of time or space.

Examples:

How many customers arrive in one hour?

How many calls are received in 10 minutes?

Formula
$$ P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!} $$

Where:

\(k\) = number of events
\(\lambda\) = average number of events
\(e\) ≈ 2.718
Example

A call center receives an average of 5 calls per hour.

What is the probability of receiving exactly 3 calls?

$$ \lambda=5,\quad k=3 $$ $$ P(X=3)=\frac{e^{-5}5^3}{3!} $$

Approximately:

$$ P(X=3)=0.1404 $$

So there is about a 14.04% probability.

#### Where is it used?
* Number of customers arriving
* Number of phone calls
* Website requests
* Number of accidents
* Number of machine failures
* Number of defects
#### AI/ML Example

In a production system, suppose an AI API receives an average of 100 requests per minute.

Poisson distribution can model:

* Number of API requests
during a particular minute

This can help with capacity planning and anomaly detection.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_value = 5

x = np.arange(0, 16)
probability = poisson.pmf(x, lambda_value)

plt.bar(x, probability)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

### Uniform Distribution
In a Uniform distribution, every value within a specified range has an equal probability.

For continuous uniform distribution:

$$ X \sim U(a,b) $$
Formula
$$ f(x)=\frac{1}{b-a} $$

for:

$$ a\leq x\leq b $$

Where:

\(a\) = minimum value
\(b\) = maximum value
Example

Suppose a bus arrives randomly between 8:00 AM and 8:30 AM.

If every time within that 30-minute interval is equally likely, the arrival time can be modeled using a uniform distribution.

Where is it used?
Random number generation
Simulation
Random sampling
Game simulations
Monte Carlo methods
AI/ML Example

Uniform distribution is commonly used for random initialization and sampling.

For example:

np.random.uniform(0, 1, 1000)

generates 1000 random values between 0 and 1.

It can also be used in hyperparameter search, where a parameter is sampled randomly from a specified range.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 10000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()

#The histogram should look approximately flat.
```

#Normal Distribution
The Normal distribution is a continuous probability distribution with a characteristic bell-shaped curve.

It is symmetric around the mean.

$$ X\sim N(\mu,\sigma^2) $$
Formula
$$ f(x)= \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$

Where:

\(\mu\) = mean
\(\sigma\) = standard deviation
\(\sigma^2\) = variance
Example

Suppose the heights of a group of people have:

Mean = 170 cm
Standard deviation = 10 cm

Most people will have heights close to 170 cm.

Empirical Rule
For a normal distribution:

| Range       | Approx. Data |
| ----------- | -----------: |
| Mean ± 1 SD |          68% |
| Mean ± 2 SD |          95% |
| Mean ± 3 SD |        99.7% |


### AI/ML Example

Normal distributions are important in ML for:

* Feature analysis
* Noise modeling
* Weight initialization
* Gaussian models
* Standardization
* Statistical assumptions

For example, z-score standardization is:

$$ z=\frac{x-\mu}{\sigma} $$



```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(
    loc=100,
    scale=10,
    size=10000
)

plt.hist(data, bins=50, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### Exponential Distribution

The Exponential distribution is commonly used to model the waiting time until the next event occurs.

Think:

Poisson → How many events?

Exponential → How long until the next event?

Formula
$$ f(x)=\lambda e^{-\lambda x} $$

where:

$$ x\geq0 $$

and \(\lambda\) is the rate of events.

Example

A customer arrives at a shop every 10 minutes on average.

Average waiting time:

$$ 10\text{ minutes} $$

Therefore:

$$ \lambda=\frac{1}{10}=0.1 $$

The exponential distribution can model the waiting time until the next customer arrives.

#### Where is it used?
* Waiting time
* Time between customer arrivals
* Machine failure time
* Service systems
* Network requests
* Queueing systems
AI/ML Example

Suppose an ML service receives requests at a constant average rate.
